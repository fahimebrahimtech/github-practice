```python
# My first data engineering practice

numbers = [10, 20, 30, 40, 50]

total = sum(numbers)
average = total / len(numbers)
maximum = max(numbers)
minimum = min(numbers)

print("Dataset:", numbers)
print("Total:", total)
print("Average:", average)
print("Maximum:", maximum)
print("Minimum:", minimum)
```

```python
## Feature Development: Data Statistics

"""This section demonstrates basic statistical
calculations using Python."""

# Feature: Data statistics

student_marks = [75, 82, 90, 68, 95, 88]

average_marks = sum(student_marks) / len(student_marks)

print("Student Marks:", student_marks)
print("Average Marks:", average_marks)
print("Highest Marks:", max(student_marks))
print("Lowest Marks:", min(student_marks))
```
